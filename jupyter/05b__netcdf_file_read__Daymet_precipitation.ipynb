# Test: reading Daymet gridded precipitation

This test is designed to read a small number of days of __Daymet 4__ data and compare the raw Daymet gridded data to the values that end up in the SWB model. There are many ways that incorrect values can be read from netCDF and make their way into a SWB calculation. Therefore, it is important to verify that the Daymet grid is being read in correctly and that the values from Daymet end up in the proper SWB grid cell.

This notebook is organized into two sections: 'test setup' and 'test comparisons'. To make the comparison, a test grid centered on the state of Minnesota is created, and SWB is run for a 14-day period. Only a short simulation is needed to verify that the Daymet netCDF file contents are read in and distributed to the proper SWB gridcells.

The strategy for comparison is straighforward. Pyproj is used to establish coordinate reference systems (CRS) for the Daymet dataset and the SWB grid. Daymet uses a Lambert Conformal Conic projection, while the SWB grid uses an Albert Equal Area projection (EPSG 5070). First, a random sample of x and y coordinates is pulled from the SWB grid. Values corresponding to the SWB coordinates are pulled from the SWB output grid. The SWB grid coordinates are transformed using Pyproj from the SWB grid CRS to the Daymet grid CRS. The values nearest to the generated Daymet coordinates are then extracted from the Daymet grid. If the SWB code is properly performing a nearest-neighbor search, the SWB grid results (as read in) and the Daymet grid results (extracted as described) should be identical.

#### Make a new swb2 model run for an example area

In the following step, a new swb2 model run is started using a pre-configured swb control file. The test case is centered on the state of Minnesota. The swb model contains 620 cells in the x direction and 688 cells in the y direction. For this test, a tiny subset of the Daymet weather data is being used. The Daymet grids have been subsetted to cover a spatial extent slightly larger than the extent of the SWB grid; in addition, only 14 timesteps are retained in order to make the comparison files as small as possible.

```python
dry_run = False
lookup_dir_arg_text = f"--lookup_dir={str(test_data_dir)}"
weather_data_dir_arg_text = f"--weather_data_dir={str(test_data_dir)}"
data_dir_arg_text = f"--data_dir={str(test_data_dir)}"
output_dir_arg_text = f"--output_dir={str(output_dir)}"
logfile_dir_arg_text = f"--logfile_dir={str(logfile_dir)}"
control_file_path = str(test_data_dir / 'swb_control_file_daymet.ctl')
output_prefix = '--output_prefix=swb_daymet_test_'

swb_arg_text = [swb2_executable, output_dir_arg_text, lookup_dir_arg_text, weather_data_dir_arg_text, 
                data_dir_arg_text, output_prefix, logfile_dir_arg_text, control_file_path]

f = open("stdout.txt", "w")

if not dry_run:
  with rm.cd(pl.Path(base_dir) / test_dir):
#    p = subprocess.Popen(swb_arg_text, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
    p = subprocess.Popen(swb_arg_text, stdout=f, stderr=subprocess.DEVNULL)
    p.wait()
```

```python
# define CRS for Daymet and SWB grids
crs_dm = pp.CRS.from_proj4(
    '+proj=lcc +lat_1=25.0 +lat_2=60.0 +lat_0=42.5 +lon_0=-100.0 '
    '+x_0=0.0 +y_0=0.0 +ellps=GRS80 +towgs84=0,0,0,0,0,0,0 +units=m'
)

# EPSG 5070 uses NAD83; the NAD83-to-WGS84 datum shift requires PROJ grid files
# (proj-data package). Including explicit +towgs84=0,0,0 makes this robust even
# when the grids are absent. The NAD83/WGS84 difference is sub-meter, which is
# irrelevant for nearest-neighbor matching on a ~4 km grid.
crs_swb = pp.CRS.from_proj4(
    '+proj=aea +lat_0=23 +lon_0=-96 +lat_1=29.5 +lat_2=45.5 '
    '+x_0=0 +y_0=0 +ellps=GRS80 +towgs84=0,0,0,0,0,0,0 +units=m'
)
```

```python
df_dm = xr.open_dataset(test_data_dir / 'daymet_v4_daily_na_prcp_2000_mn__14-day.nc')
df_swb = xr.open_dataset( pl.Path(base_dir).resolve() / test_dir / output_dir / 
                         "swb_daymet_test_gross_precipitation__2000-01-01_to_2000-01-14__688_by_620.nc")

da_dm = df_dm.to_dataarray()
da_swb = df_swb.to_dataarray()
```

```python
# use pyproj to determine the coordinate values in the Daymet data set based on the subset pulled from the SWB grid
proj = pp.transformer.Transformer.from_crs(crs_swb, crs_dm)
```

```python
def get_random_val(drange):
    v = rnd.randrange(start=int(drange*.1), stop=int(drange*.9))
    return v
```

#### Generate a set of coordinate values for data extraction and comparison

In this step, a random set of xy values is drawn from the SWB grid. These are transformed into another set of xy values that specifies the nearest location in terms of the Daymet CRS. 

```python
number_of_xy_pairs = 30
xrange = df_swb.sizes['x']
yrange = df_swb.sizes['y']
x_indices = [get_random_val(xrange) for i in range(number_of_xy_pairs)]
y_indices = [get_random_val(yrange) for i in range(number_of_xy_pairs)]
x_swb = da_swb.x.values[x_indices]
y_swb = da_swb.y.values[y_indices]
x_dm, y_dm = proj.transform(x_swb, y_swb)

# Hard fail if the coordinate transform returns inf or NaN.
# This typically means the PROJ datum grid files are missing.
# Fix: install the 'proj-data' package (conda-forge).
assert np.all(np.isfinite(x_dm)) and np.all(np.isfinite(y_dm)), (
    'Coordinate transform returned non-finite values (inf or NaN). '
    'This usually means PROJ datum grid files are not installed. '
    'Install them with: mamba install -c conda-forge proj-data'
)
```

#### Create subset of SWB and Daymet values

In this step, the values contained in the SWB output file are extracted for the subset of SWB xy values. The corresponding values are also extracted from the Daymet grid files using the subset of Daymet grid xy values. The extracted values are stored in separate lists.

```python
swb_vals = []
daymet_vals = []

for i in range(number_of_xy_pairs):
    data_dm = df_dm.sel(x=x_dm[i], y=y_dm[i], method='nearest')
    data_swb = df_swb.sel(x=x_swb[i],y=y_swb[i], method='nearest')
    pdm = list(data_dm.prcp.values)
    pswb = list(data_swb.gross_precipitation.values*25.4)
    for item in pswb:
        swb_vals.append(item)
    for item in pdm:
        daymet_vals.append(item)
    
```

#### Create Pandas dataframe from lists

```python
results_df = pd.DataFrame(data={'swb_precip': swb_vals,'daymet_precip': daymet_vals})
```

#### Drop NaN values from the dataframe of results

Generally, there are many areas within a SWB grid that might be inactive. These areas might be over open water, or cover parts of the model domain for which no data exist. Here we drop all rows from the dataframe that contain 'NaN' values. 

```python
# dropping 'NaN' values, since these likely exist owing to inactive model cells within the SWB grid
results_df.dropna(axis=0, inplace=True)
```

## Test comparisons

```python
fig, ax = plt.subplots(1,1,figsize=(12, 8),sharex=False)
ax.plot(results_df['swb_precip'], results_df['daymet_precip'], marker='.', linestyle=' ', color='blue', markersize=4)
ax.plot(range(13), range(13), linestyle='-', color='red', label='1:1 line')
ax.set_ylabel('Daymet precipitation, in millimeters')
ax.set_xlabel('SWB precipitation, in millimeters')
ax.set_title('Precipitation comparison: raw Daymet value compared with value assigned to SWB cell')
ax.legend()
```
